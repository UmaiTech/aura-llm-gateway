# Feedback → adaptive few-shot learning

The gateway stores thumbs-up/down feedback per response. Approved samples become
candidates for few-shot injection into later contexts, so the gateway gets better at
*your* task over time.

- `POST /v1/feedback` — record a rating
- `GET /v1/feedback` — list samples (for few-shot injection)
- `GET /v1/feedback/stats` — aggregate counts

```python
from aura import AuraClient, FeedbackSignal

client = AuraClient()
```

```python
# 1. Create a response you want to rate
response = client.responses.create(
    model="gpt-5.4-mini",
    input="Summarize the waterfall model in one sentence.",
)
print(f"rated response: {response.id}")
print(f"answer:         {response.output_text}")
```

```python
# 2. Submit a thumbs-up with a reason + tags
result = client.feedback.submit(
    response_id=response.id,
    signal=FeedbackSignal.THUMBS_UP,
    reason="concise and accurate",
    tags=["summarization"],
)
print(result)
```

```python
# 3. List samples (what the gateway can few-shot from)
samples = client.feedback.list()
print(f"total samples: {samples.get('total')}")
for sample in samples.get("samples", []):
    print(sample)
```

```python
# 4. Aggregate stats
print(client.feedback.stats())
```

Approved samples are sampled into later request contexts automatically when the gateway
is configured for adaptive few-shot (see the feedback docs in `crates/aura-core`).
`DELETE /v1/feedback/{id}` lets you remove a bad sample.
