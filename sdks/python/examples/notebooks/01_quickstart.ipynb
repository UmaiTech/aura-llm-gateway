# Aura LLM Gateway — Quickstart

A 5-minute *hello world* against the [Open Responses API](https://www.openresponses.org/specification) through the Aura gateway.

## Prerequisites

1. A running gateway (default `http://localhost:8080`).
2. An API key, either passed to `AuraClient(api_key=...)` or set as `AURA_API_KEY`.
3. The SDK: `pip install aura-llm==0.17.0` (or run from `sdks/python` with `pip install -e .`).

> Aura proxies multiple providers behind one endpoint, so `model` can be any model your gateway is configured for — e.g. `gpt-5.4-mini`, `claude-sonnet-4-6`, or a Together slug.

```python
# 1. Install (skip if already installed)
!pip install -q aura-llm==0.17.0
```

```python
# 2. Client
from aura import AuraClient

# Uses AURA_BASE_URL / AURA_API_KEY env vars when not passed.
client = AuraClient()
print(f"gateway: {client.base_url}")
```

```python
# 3. First completion
response = client.responses.create(
    model="gpt-5.4-mini",
    input="What is 2 + 2? Answer in one sentence.",
)

print(f"id:     {response.id}")
print(f"model:  {response.model}")
print(f"status: {response.status}")
print(f"answer: {response.output_text}")
```

```python
# 4. See usage + cost attribution
if response.usage:
    u = response.usage
    print(f"tokens: {u.input_tokens} in / {u.output_tokens} out")
    if u.cost_usd is not None:
        print(f"cost:   ${u.cost_usd:.6f}")
```
